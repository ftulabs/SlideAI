# L10 Neural Networks · One neuron
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
x = torch.tensor([2.0, 4.0])
w = torch.tensor([0.5, -0.3]); b = torch.tensor(0.1)
z = w @ x + b                  # weighted sum + bias
a = torch.sigmoid(z)           # squash
print(z.item(), a.item())      # -0.1, 0.475
```
