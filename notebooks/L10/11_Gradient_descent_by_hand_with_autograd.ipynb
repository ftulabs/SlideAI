# L10 Neural Networks · Gradient descent by hand with autograd
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
w = torch.tensor(0.0, requires_grad=True)
eta = 0.1
for t in range(50):
    J = (3 * w - 5) ** 2           # minimum at w = 5/3
    J.backward()
    with torch.no_grad():
        w -= eta * w.grad          # one step downhill
        w.grad.zero_()
print(w.item())                    # about 1.667
```
