# L10 Neural Networks · LSTM
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
lstm = nn.LSTM(input_size=10, hidden_size=20, batch_first=True)
out, (h, c) = lstm(torch.randn(4, 7, 10))
print(out.shape, c.shape)
print(sum(p.numel() for p in lstm.parameters()))   # 4 x 640 = 2560 (PyTorch keeps two bias vectors)
```
