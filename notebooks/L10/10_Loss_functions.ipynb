# L10 Neural Networks · Loss functions
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
y_hat = torch.tensor([2.5, 0.0]); y = torch.tensor([3.0, -0.5])
print(nn.MSELoss()(y_hat, y))                         # regression
logits = torch.tensor([[2.0, 1.0, 0.0]]); cls = torch.tensor([0])
print(nn.CrossEntropyLoss()(logits, cls))             # = -log softmax(logits)[0] = 0.408
print(nn.BCEWithLogitsLoss()(torch.tensor([0.3]), torch.tensor([1.0])))   # binary
```
