# L10 Neural Networks · Fraud net in NumPy
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import numpy as np

def relu(z):    return np.maximum(0, z)
def sigmoid(z): return 1 / (1 + np.exp(-z))

def forward(X, W1, b1, W2, b2):
    a1 = relu(X @ W1 + b1)       # (m, 3)  hidden layer
    p  = sigmoid(a1 @ W2 + b2)   # (m,)    P(fraud)
    return a1, p

W1 = np.array([[1.5, 0.0, 1.0],
               [0.0, 0.8, 0.5]])
b1 = np.array([-2.0, -2.4, -3.0])
W2 = np.array([2.0, 2.5, 1.5]); b2 = -2.0
X  = np.array([[0.2, 1.0], [2.4, 2.0],
               [0.6, 5.0], [2.6, 6.0]])
print(forward(X, W1, b1, W2, b2)[1])
# [0.119 0.858 0.896 1.   ]
```
