# L10 Neural Networks · Forward pass, layer by layer
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
W1 = torch.tensor([[0.8, -0.5], [0.3, 0.9], [-0.6, 0.4]]); b1 = torch.tensor([0.1, -0.2, 0.05])
W2 = torch.randn(2, 3); b2 = torch.zeros(2)
x = torch.tensor([1.0, 2.0])
z1 = W1 @ x + b1;  a1 = torch.relu(z1)
z2 = W2 @ a1 + b2; p = torch.softmax(z2, dim=0)
print(z1, a1, p, sep="\n")
```
