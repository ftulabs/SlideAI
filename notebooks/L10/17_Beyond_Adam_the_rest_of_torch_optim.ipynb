# L10 Neural Networks · Beyond Adam: the rest of torch.optim
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
p = [torch.nn.Parameter(torch.randn(4, 4))]
O = torch.optim
for name, opt in {"NAdam": O.NAdam(p, lr=2e-3), "RAdam": O.RAdam(p, lr=1e-3),
                  "AMSGrad": O.Adam(p, lr=1e-3, amsgrad=True), "Adamax": O.Adamax(p, lr=2e-3),
                  "Adadelta": O.Adadelta(p), "ASGD": O.ASGD(p, lr=0.01)}.items():
    loss = (p[0] ** 2).sum(); opt.zero_grad(); loss.backward(); opt.step(); print(name, "ok")
lbfgs = O.LBFGS(p, lr=1)                  # quasi-Newton: needs a closure
def closure():
    lbfgs.zero_grad(); l = (p[0] ** 2).sum(); l.backward(); return l
lbfgs.step(closure); print("LBFGS", round((p[0] ** 2).sum().item(), 6))
```
