# L10 Neural Networks · Activation functions and their derivatives
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
z = torch.tensor([-2.0, 0.5], requires_grad=True)
for f in [torch.sigmoid, torch.tanh, torch.relu, lambda t: torch.nn.functional.leaky_relu(t, 0.1)]:
    z.grad = None
    f(z).sum().backward()               # autograd gives the derivative at each z
    print(z.grad.tolist())
```
