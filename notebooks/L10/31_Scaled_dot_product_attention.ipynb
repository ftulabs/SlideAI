# L10 Neural Networks · Scaled dot-product attention
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn.functional as F
Q = torch.randn(1, 5, 8); K = torch.randn(1, 5, 8); V = torch.randn(1, 5, 8)
A = torch.softmax(Q @ K.transpose(-2, -1) / 8 ** 0.5, dim=-1)
out = A @ V
print(torch.allclose(out, F.scaled_dot_product_attention(Q, K, V), atol=1e-6))
```
