# L10 Neural Networks · A feed-forward network
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
model = nn.Sequential(
    nn.Linear(2, 4), nn.ReLU(),     # hidden layer 1
    nn.Linear(4, 4), nn.ReLU(),     # hidden layer 2
    nn.Linear(4, 1), nn.Sigmoid())  # output P(class 1)
print(model)
print("parameters:", sum(p.numel() for p in model.parameters()))   # 12 + 20 + 5 = 37
print(model(torch.tensor([[1.0, 2.0]])))
```
