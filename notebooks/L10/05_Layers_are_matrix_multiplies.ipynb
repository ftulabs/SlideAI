# L10 Neural Networks · Layers are matrix multiplies
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
layer = nn.Linear(3, 4)                 # weight (4, 3), bias (4,)
X = torch.randn(32, 3)                  # a batch of m = 32 rows
Z = layer(X)                            # = X @ layer.weight.T + layer.bias
print(layer.weight.shape, Z.shape)      # (4, 3)  (32, 4)
print(torch.allclose(Z, X @ layer.weight.T + layer.bias))
```
