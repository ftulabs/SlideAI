# L10 Neural Networks · 2-D convolution
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
conv = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, stride=1, padding=1)
x = torch.randn(8, 3, 32, 32)                  # batch, channels, height, width
print(conv(x).shape)                           # (8, 16, 32, 32)
print(conv.weight.shape, sum(p.numel() for p in conv.parameters()))   # (16,3,3,3), 448
```
