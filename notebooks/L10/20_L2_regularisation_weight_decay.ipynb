# L10 Neural Networks · L2 regularisation = weight decay
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
model = nn.Linear(10, 1)
opt = torch.optim.SGD(model.parameters(), lr=0.1, weight_decay=1e-4)   # adds lambda*w to the gradient
# or explicitly:  loss = loss + 1e-4 / 2 * sum((p ** 2).sum() for p in model.parameters())
x, y = torch.randn(4, 10), torch.randn(4, 1)
loss = nn.MSELoss()(model(x), y); opt.zero_grad(); loss.backward(); opt.step(); print("ok")
```
