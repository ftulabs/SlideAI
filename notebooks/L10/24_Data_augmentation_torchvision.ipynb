Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
# needs torchvision (preinstalled on Colab)
import torch
from torchvision import transforms as T
aug = T.Compose([T.RandomHorizontalFlip(), T.RandomResizedCrop(32, scale=(0.8, 1.0)),
                 T.ColorJitter(0.2, 0.2)])
img = torch.rand(3, 40, 40)
print(aug(img).shape)            # a new random view every call, same label
```
