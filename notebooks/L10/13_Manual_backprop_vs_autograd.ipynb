# L10 Neural Networks · Manual backprop vs autograd
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
torch.manual_seed(0)
X = torch.randn(8, 2); y = torch.randn(8, 1)
W1 = torch.randn(2, 3, requires_grad=True); W2 = torch.randn(3, 1, requires_grad=True)
H = torch.tanh(X @ W1); Y = H @ W2
L = ((Y - y) ** 2).mean(); L.backward()
with torch.no_grad():                         # the same gradients by hand
    d2 = 2 * (Y - y) / len(X)                 # dL/dY
    gW2 = H.T @ d2
    d1 = (d2 @ W2.T) * (1 - H ** 2)           # back through W2 and tanh'
    gW1 = X.T @ d1
print(torch.allclose(gW1, W1.grad), torch.allclose(gW2, W2.grad))
```
