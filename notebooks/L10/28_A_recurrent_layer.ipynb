# L10 Neural Networks · A recurrent layer
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
rnn = nn.RNN(input_size=10, hidden_size=20, batch_first=True)
x = torch.randn(4, 7, 10)                 # 4 sequences, 7 steps, 10 features
out, h_T = rnn(x)
print(out.shape, h_T.shape)               # (4, 7, 20)  (1, 4, 20)
print(sum(p.numel() for p in rnn.parameters()))   # 20*(20+10) + 2*20 = 640
```
