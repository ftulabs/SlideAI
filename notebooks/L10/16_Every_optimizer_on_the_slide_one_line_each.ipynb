# L10 Neural Networks · Every optimizer on the slide, one line each
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
p = [torch.nn.Parameter(torch.randn(3))]
O = torch.optim
opts = {"SGD": O.SGD(p, lr=0.1),
        "Momentum": O.SGD(p, lr=0.1, momentum=0.9),
        "Nesterov": O.SGD(p, lr=0.1, momentum=0.9, nesterov=True),
        "AdaGrad": O.Adagrad(p, lr=0.1),
        "RMSProp": O.RMSprop(p, lr=0.01, alpha=0.9),
        "Adam": O.Adam(p, lr=1e-3, betas=(0.9, 0.999)),
        "AdamW": O.AdamW(p, lr=1e-3, weight_decay=0.01)}
for name, opt in opts.items():
    loss = (p[0] ** 2).sum(); opt.zero_grad(); loss.backward(); opt.step()
    print(name, "ok")
```
