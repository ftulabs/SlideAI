# L10 Neural Networks · A small CNN classifier
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
cnn = nn.Sequential(
    nn.Conv2d(3, 16, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),    # 32 -> 16
    nn.Conv2d(16, 32, 3, padding=1), nn.ReLU(), nn.MaxPool2d(2),   # 16 -> 8
    nn.Flatten(), nn.Linear(32 * 8 * 8, 10))
print(cnn(torch.randn(4, 3, 32, 32)).shape)                        # (4, 10)
print(sum(p.numel() for p in cnn.parameters()))
```
