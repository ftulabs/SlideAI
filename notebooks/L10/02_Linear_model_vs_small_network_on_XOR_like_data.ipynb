# L10 Neural Networks · Linear model vs small network on XOR-like data
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
torch.manual_seed(0)
X = torch.rand(400, 2) * 2 - 1
y = ((X[:, 0] * X[:, 1]) < 0).float()          # XOR: opposite signs
def fit(model, steps=2000):
    opt = torch.optim.Adam(model.parameters(), lr=0.02)
    for _ in range(steps):
        loss = nn.functional.binary_cross_entropy_with_logits(model(X).squeeze(1), y)
        opt.zero_grad(); loss.backward(); opt.step()
    return ((model(X).squeeze(1) > 0).float() == y).float().mean().item()
linear = nn.Linear(2, 1)                                        # logistic regression
deep_but_linear = nn.Sequential(nn.Linear(2, 16), nn.Linear(16, 1))   # no activation
mlp = nn.Sequential(nn.Linear(2, 16), nn.Tanh(), nn.Linear(16, 1))
for name, m in [("linear", linear), ("2 linear layers", deep_but_linear), ("MLP", mlp)]:
    print(f"{name:16s} accuracy {fit(m):.2f}")
```
