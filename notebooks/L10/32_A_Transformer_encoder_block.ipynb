# L10 Neural Networks · A Transformer encoder block
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
block = nn.TransformerEncoderLayer(d_model=64, nhead=4, dim_feedforward=256, batch_first=True, norm_first=True)
x = torch.randn(2, 10, 64)                    # 2 sentences, 10 tokens, d = 64
print(block(x).shape)
print(sum(p.numel() for p in block.parameters()))   # 49,984
```
