# L10 Neural Networks · Batch normalisation
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
bn = nn.BatchNorm1d(3)
x = torch.randn(64, 3) * 5 + 10
y = bn(x)                                   # training mode: batch statistics
print(y.mean(0), y.std(0))                  # about 0 and 1
print(bn.running_mean)                      # used later in eval mode
```
