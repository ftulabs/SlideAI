# L10 Neural Networks · Xavier and He initialisation
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
layer = nn.Linear(512, 512)
nn.init.kaiming_normal_(layer.weight, nonlinearity="relu")   # He: std = sqrt(2/512)
print(layer.weight.std().item())                            # about 0.0625
nn.init.xavier_uniform_(layer.weight)                       # Xavier/Glorot for tanh
print(layer.weight.std().item())                            # about sqrt(1/512) = 0.044
```
