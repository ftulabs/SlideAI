# L10 Neural Networks · Same forward pass as an nn.Module
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
class FraudNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 3)
        self.fc2 = nn.Linear(3, 1)
    def forward(self, x):
        a1 = torch.relu(self.fc1(x))
        return torch.sigmoid(self.fc2(a1))
net = FraudNet()
with torch.no_grad():                              # load the bank's trained weights
    # nn.Linear stores weight as (out, in) = W1 transposed
    net.fc1.weight.copy_(torch.tensor([[1.5, 0.0],
                                       [0.0, 0.8],
                                       [1.0, 0.5]]))
    net.fc1.bias.copy_(torch.tensor([-2.0, -2.4, -3.0]))
    net.fc2.weight.copy_(torch.tensor([[2.0, 2.5, 1.5]]))
    net.fc2.bias.fill_(-2.0)
X = torch.tensor([[0.2, 1.0], [2.4, 2.0],
                  [0.6, 5.0], [2.6, 6.0]])
print(net(X).squeeze(1))   # 0.119 0.858 0.896 1.000
```
