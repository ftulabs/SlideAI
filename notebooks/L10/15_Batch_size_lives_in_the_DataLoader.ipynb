# L10 Neural Networks · Batch size lives in the DataLoader
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
from torch.utils.data import TensorDataset, DataLoader
ds = TensorDataset(torch.randn(48, 1), torch.randn(48, 1))
for B in [48, 1, 8]:                      # full batch, stochastic, mini-batch
    loader = DataLoader(ds, batch_size=B, shuffle=True)
    print("B =", B, "->", len(loader), "steps per epoch")
```
