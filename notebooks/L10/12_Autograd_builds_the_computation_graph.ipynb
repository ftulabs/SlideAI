# L10 Neural Networks · Autograd builds the computation graph
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch
x, y = torch.tensor(1.0), torch.tensor(0.7)
w1 = torch.tensor(0.5, requires_grad=True); w2 = torch.tensor(2.0, requires_grad=True)
h = torch.sigmoid(w1 * x)          # forward: every op is recorded
y_hat = w2 * h
L = 0.5 * (y_hat - y) ** 2
print(L.grad_fn)                   # last node of the graph
L.backward()                       # chain rule, back through the graph
print(w1.grad.item(), w2.grad.item())
```
