# L10 Neural Networks · Dropout
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
drop = nn.Dropout(p=0.5)
a = torch.ones(8)
drop.train(); print(drop(a))     # about half zeroed, the rest scaled by 2
drop.eval();  print(drop(a))     # test time: identity
```
