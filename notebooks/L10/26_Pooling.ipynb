# L10 Neural Networks · Pooling
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
x = torch.tensor([[[[1., 3, 2, 1], [4, 6, 5, 0], [7, 2, 1, 8], [3, 1, 0, 9]]]])
print(nn.MaxPool2d(2)(x))        # [[6, 5], [7, 9]]
print(nn.AvgPool2d(2)(x))
```
