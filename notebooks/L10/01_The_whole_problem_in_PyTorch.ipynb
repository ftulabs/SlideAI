# L10 Neural Networks · The whole problem in PyTorch
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import torch, torch.nn as nn
X = torch.randn(256, 784)                 # n = 256 images, d = 784 pixels
y = torch.randint(0, 10, (256,))          # labels 0..9
f = nn.Sequential(nn.Linear(784, 128), nn.ReLU(), nn.Linear(128, 10))   # f_theta
L = nn.CrossEntropyLoss()                 # L(y_hat, y)
opt = torch.optim.AdamW(f.parameters(), lr=1e-3)
for step in range(100):                   # arg min over theta, step by step
    loss = L(f(X), y)
    opt.zero_grad(); loss.backward(); opt.step()
print("training loss", round(loss.item(), 3))
```
