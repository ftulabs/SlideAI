# L10 Neural Networks · Early stopping
Run the cell below (Shift+Enter). Colab already has PyTorch installed.

```python
import copy
best, best_state, patience, wait = float("inf"), None, 5, 0
val_losses = [0.9, 0.7, 0.6, 0.55, 0.56, 0.57, 0.58, 0.60, 0.61, 0.62]   # one per epoch
for epoch, v in enumerate(val_losses):
    if v < best: best, wait, best_state = v, 0, copy.deepcopy({"epoch": epoch})   # model.state_dict()
    else:
        wait += 1
        if wait >= patience: break
print("stopped at", epoch, "best epoch", best_state["epoch"], "val", best)
```
